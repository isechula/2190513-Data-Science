```python
#for submission

Q1 = lambda: len(pandas.read_csv('/data/videos.csv').drop_duplicates())

Q2 = lambda df : len(df.loc[df['dislikes'] > df["likes"], ['title']].groupby('title'))

Q3 = lambda df: len(df.loc[df['trending_date'] == '18.22.01', ['comment_count']].loc[df["comment_count"] > 10000, ['comment_count']])

Q4 = lambda df : df.groupby("trending_date")['comment_count'].mean().idxmin()

def Q5(vdo_df):
    with open("/data/category_id.json") as f:
        category_data = json.load(f)

    category_sports = list(filter(lambda x: x["snippet"]["title"] == "Sports", category_data["items"]))
    category_comedy = list(filter(lambda x: x["snippet"]["title"] == "Comedy", category_data["items"]))

    sports_views = vdo_df[vdo_df["category_id"].isin([int(item["id"]) for item in category_sports])].groupby("trending_date")["views"].sum()
    comedy_views = vdo_df[vdo_df["category_id"].isin([int(item["id"]) for item in category_comedy])].groupby("trending_date")["views"].sum()
    return (sports_views > comedy_views).sum()
```
